A2 — Analogy and Similarity Ranking - 6 points

```python
import numpy as np

A = np.array([
    [1, 1, 0, 0],
    [0, 1, 1, 0],
    [0, 0, 1, 0],
    [1, 0, 0, 1]
])

x_France  = np.array([1, 0, 0, 1])
x_Italy   = np.array([1, 1, 0, 0])
x_Paris   = np.array([1, 0, 1, 1])
x_Rome    = np.array([1, 1, 1, 0])
x_Berlin  = np.array([0, 1, 1, 1])
x_Madrid  = np.array([1, 0, 0, 0])

def embed(x):
    return A @ x

def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

# Compute embeddings
y_France = embed(x_France)
y_Italy = embed(x_Italy)
y_Paris = embed(x_Paris)
y_Rome = embed(x_Rome)
y_Berlin = embed(x_Berlin)
y_Madrid = embed(x_Madrid)

# Compute q
q = y_Rome - y_Italy + y_France

print("q =", q)

# Capital embeddings
capitals = {
    "Paris": y_Paris,
    "Rome": y_Rome,
    "Berlin": y_Berlin,
    "Madrid": y_Madrid
}

# Compute cosine similarities
scores = {}
for name, vec in capitals.items():
    scores[name] = cosine_similarity(q, vec)

# Rank capitals
ranking = sorted(scores.items(), key=lambda x: x[1], reverse=True)

print("\nCosine similarities:")
for name, score in scores.items():
    print(f"{name}: {score:.3f}")

print("\nRanking:")
for i, (name, score) in enumerate(ranking, start=1):
    print(f"{i}. {name} -> {score:.3f}")
```

A3 — Rank and Semantic Diversity - 8 points

```python
import numpy as np

Y = np.array([
    [1, 2, 1, 0],
    [0, 1, 2, 1],
    [0, 0, 1, 1],
    [2, 1, 0, 1]
])

rank_Y = np.linalg.matrix_rank(Y)

print("Y =")
print(Y)
print("\nrank(Y) =", rank_Y)

if rank_Y == Y.shape[1]:
    print("The columns are linearly independent.")
else:
    print("The columns are not linearly independent.")
```
