B1 — Compute the SVD components - 14 points 

```python
import numpy as np

# Define the centered data matrix X
X = np.array([
    [-2, -1],
    [-1, -2],
    [ 0,  0],
    [ 1,  2],
    [ 2,  1]
])

# Compute SVD using NumPy
# Note: Vt is V-transpose
U, s, Vt = np.linalg.svd(X, full_matrices=True)
V = Vt.T 

print("--- B1: SVD Verification ---")
print(f"Singular Values (sigma):\n{s}")
print(f"Matrix V (Right Singular Vectors):\n{V}")
print(f"Matrix U (Left Singular Vectors):\n{U}")
```

B2 — Compress the data with a rank-1 approximation - 6 points

```python
# Extract the first singular value, first column of U, and first column of V
sigma1 = s[0]
u1 = U[:, 0].reshape(-1, 1)
v1 = V[:, 0].reshape(-1, 1)

# Compute the Rank-1 approximation
X1 = sigma1 * (u1 @ v1.T)

# Calculate the reconstruction error using the Frobenius norm
reconstruction_error = np.linalg.norm(X - X1)

# Calculate the fraction of variance retained
variance_retained = (s[0]**2) / np.sum(s**2)

print("--- B2: Compression Results ---")
print(f"Rank-1 Approximation Matrix X1:\n{X1}")
print(f"Reconstruction Error: {reconstruction_error:.4f}")
print(f"Variance Retained: {variance_retained:.2%}")
```

B3 — PCA: going from 2D to 1D - 6 points

```python
# Project data points onto the first principal component (v1)
# This gives the 1D coordinates for each 2D point
z = X @ V[:, 0]

print("--- B3: PCA 1D Coordinates ---")
for i, coordinate in enumerate(z):
    print(f"Point {i+1} coordinate: {coordinate:.4f}")
```
